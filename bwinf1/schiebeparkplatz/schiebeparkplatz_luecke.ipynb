Jede Datei beschreibt einen Schiebeparkplatz und enthält

  * in der ersten Zeile die Bezeichnung des ersten und letzten Autos
    auf den normalen Parkplätzen (Bezeichnungen „A“, „B“, „C“, usw.),
  * in der zweiten Zeile die Anzahl n der quer stehenden Autos und
  * in den weiteren n Zeilen für jedes quer stehende Auto jeweils.
    * die Bezeichnung des quer stehenden Autos und
    * die Position $p_i$ des quer stehenden Autos.

Die Position startet links von 0. Jedes verschiebbare Auto nimmt wie in
der Zeichnung gezeigt zwei Positionen ein (also jeweils $p_i$ und $p_{i + 1}$).

Zu einem gegebenen Auto oben unterscheiden wir drei Fälle:
- (1) unten ist frei
- (2) unten ist auto, das sich links fortsetzt (z.B. D)
- (3) unten ist auto, das sich rechts fortsetzt (z.B. C)

#### Pseudocode für unsere Lösung

Gegeben ist Auto x bei Index k
f = Fall, der für Auto x vorliegt (1-3)
Falls f == 1:
    nichts zu tun, Auto kann rausfahren
Falls f == 2:
    el = suche den index der ersten Lücke links von k
    zähle die Autos von k bis el
    zl = suche den index der zweiten Lücke rechts von k
    zahle die Autos von k bis zl
    el <= zl:
        schiebe die autos vom linkesten ausgehend nacheinander bis zu el
    sonst:
        schiebe die autos vom rechtesten ausgehend nacheinander bis zu zl
Falls f == 3:
    el = suche den index der ersten Lücke rechts von k
    zähle die Autos von k bis el
    zl = suche den index der zweiten Lücke links von k
    zahle die Autos von k bis zl
    el <= zl:
        schiebe die autos vom linkesten ausgehend nacheinander bis zu el
    sonst:
        schiebe die autos vom rechtesten ausgehend nacheinander bis zu zl 

```python
def ersteLuecke_links(unten, k):
    ''' 
    unten: Liste mit den querstehenden Autos, z.B. ['.', '.', 'H', 'H', '.', 'I', 'I']
    k: index in der Liste unten
    returns: Index j der ersten Luecke links von k, also j < k, None wenn keine vorhanden
    '''
    j = k
    while j >= 0 and unten[j] != '.':
        j = j-1
    return j if j >= 0 else None

def zweiteLuecke_links(unten, k):
    ''' 
    unten: Liste mit den querstehenden Autos, z.B. ['.', '.', 'H', 'H', '.', 'I', 'I']
    k: index in der Liste unten
    returns: Index j der zweiten Luecke links von k, also j < k, None wenn keine vorhanden
    '''
    j = ersteLuecke_links(unten, k)
    if j is not None and j > 0:
        return ersteLuecke_links(unten, j-1)    
    return None
```

```python
def ersteLuecke_rechts(unten, k):
    ''' 
    unten: Liste mit den querstehenden Autos, z.B. ['.', '.', 'H', 'H', '.', 'I', 'I']
    k: index in der Liste unten
    returns: Index j der ersten Luecke rechts von k, also k < j, None wenn keine vorhanden
    '''
    j = k
    while j < len(unten) and unten[j] != '.':
        j = j+1
    return j if j < len(unten) else None

def zweiteLuecke_rechts(unten, k):
    ''' 
    unten: Liste mit den querstehenden Autos, z.B. ['.', '.', 'H', 'H', '.', 'I', 'I']
    k: index in der Liste unten
    returns: Index j der zweiten Luecke rechts von k, also k < j, None wenn keine vorhanden
    '''
    j = ersteLuecke_rechts(unten, k)
    if j is not None and j < len(unten):
        return ersteLuecke_rechts(unten, j+1)    
    return None
```

```python
def schieb_links(unten, L, k):
    ''' alle Autos zwischen L und k werden nach links verschoben, höchstens bis L 
    '''
    moves = []
    tmp = list(unten)
    i = L
    strecke = 0
    while i <= k:
        while tmp[i] == '.' and i<=k:
            i += 1
        if i-L > 0:
            moves.append(f'{tmp[i]} um {i-L} nach links')
            strecke += (i-L)
        tmp[i] = '.'
        if i-L == 2:
            tmp[i-1] = '.'
        L = i+1
        i = L+2
        
    return moves, strecke

print(schieb_links('...HH.KK..JJ',2,10))
```

Zur Suche in dem Baum der Möglichkeiten benötigen wir die Funktionen *goaltest* und *nextstates*.

```python
raus = 2
def goaltest(state):
    ''' raus = index des Wagens in der oberen Reihe, der rausgefahren werden soll '''
    return state[raus] == '.'
```

```python
def nextstates(state):
    tmp = []
    for c, pos in quer:
        new_state = list(state)
        k = pos
        while k+2 < len(state) and state[k+2]=='.':   # solange rechts Platz ist
            new_state[k]='.'                          # verschiebe die beiden Zeichen in new_state                     
            new_state[k+2]=c                          # um eins nach rechts
            tmp.append(tuple(new_state))              # speichere den Zustand als möglichen Folgezustand
            k+=1            
        k = pos
        new_state = list(state)
        while k-1 >= 0 and new_state[k-1]=='.':       # solange links Platz ist
            new_state[k-1]=c
            new_state[k+1]='.'            
            tmp.append(tuple(new_state))
            k-=1
    return tmp
```

```python
from collections import deque
def bfs(startstate):
    frontier =  deque([startstate])
    prev = {startstate:None}
    while frontier:
        state = frontier.popleft()  
        if goaltest(state):
            return prev, state
        for v in nextstates(state):
            if v not in prev:
                frontier.append(v)
                prev[v] = state
    return None, None

def reconstructPath(prev,goalstate):
    state = goalstate
    path = []
    while state is not None:
        path.append(state)
        state = prev[state]
    path.reverse()
    return path
```
